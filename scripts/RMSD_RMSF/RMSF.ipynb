```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pre_rmsf=pd.read_csv('../../data/RMSD_RMSF/RMSF_PRO_ALL_SIMS.csv')
unb_rmsf=pd.read_csv('../../data/RMSD_RMSF/RMSF_UNB_ALL_SIMS.csv')
bnd_rmsf=pd.read_csv('../../data/RMSD_RMSF/RMSF_BND_ALL_SIMS.csv')
```

```python
plt.figure(figsize=(20,2))


plt.plot([i+1 for i in pre_rmsf.index],pre_rmsf['S1'], color='#FF7878')
plt.plot([i+1 for i in pre_rmsf.index],pre_rmsf['S2'], color='#FF7878')
plt.plot([i+1 for i in pre_rmsf.index],pre_rmsf['S3'], color='#FF7878')

plt.ylim(0,1.2)

plt.xlabel('Residues')
plt.ylabel('RMSF')

plt.savefig("../../plots/RMSD_RMSF/RMSF_pre.png",
           dpi=450, bbox_inches='tight', transparent=True)
```

```python
plt.figure(figsize=(20,2))
plt.plot([i+1 for i in unb_rmsf.index],unb_rmsf['S1'], color='#5A91FF')
plt.plot([i+1 for i in unb_rmsf.index],unb_rmsf['S2'], color='#5A91FF')
plt.plot([i+1 for i in unb_rmsf.index],unb_rmsf['S3'], color='#5A91FF')

plt.ylim(0,1.2)
plt.xlabel('Residues')
plt.ylabel('RMSF')
plt.savefig("../../plots/RMSD_RMSF/RMSF_unb.png",
           dpi=450, bbox_inches='tight', transparent=True)
```
