```python
with open ("../../../data/dihedrals/net_communication/MI_script_prep.sh","w+") as f_nc:
    for fil in os.listdir('../../../../LC3_MAIN_TRAJECTORIES/PROTEIN_ONLY_FITTED/'):
        for sub_fil in os.listdir("../../../../LC3_MAIN_TRAJECTORIES/PROTEIN_ONLY_FITTED/%s"%(fil)):
            if sub_fil.endswith('.gro') and fil.startswith("._")==False:
                fil_name=sub_fil.replace('_0ns.gro','')
                print(fil_name)
                
                
                #dihedral-dihedral MI 
                f_nc.write("cp compute_MI_cc.py %s/BB/PSI/DIHEDRALS/.\n"%fil_name)
                f_nc.write("cp compute_MI_cc.py %s/BB/PHI/DIHEDRALS/.\n"%fil_name)
                
                f_nc.write("cp compute_MI_cc.py %s/BB/PSI/DIHEDRALS/.\n"%fil_name)
                f_nc.write("cp compute_MI_cc.py %s/BB/PHI/DIHEDRALS/.\n"%fil_name)
                
                f_nc.write("cp compute_MI_cc.py %s/SC/CHI1/DIHEDRALS/.\n"%fil_name)
                
                #rotamer-rotamer MI
                
                f_nc.write("cp compute_MI_dd.py %s/BB/PSI/ROTAMERS/.\n"%fil_name)
                f_nc.write("cp compute_MI_dd.py %s/BB/PHI/ROTAMERS/.\n"%fil_name)
                
                f_nc.write("cp compute_MI_dd.py %s/BB/PSI/ROTAMERS/.\n"%fil_name)
                f_nc.write("cp compute_MI_dd.py %s/BB/PHI/ROTAMERS/.\n"%fil_name)
                
                f_nc.write("cp compute_MI_dd.py %s/SC/CHI1/ROTAMERS/.\n"%fil_name)
                
                
                #orderness-orderness MI
                
                f_nc.write("cp compute_MI_dd_orderness.py %s/BB/PSI/ORDERNESS/.\n"%fil_name)
                f_nc.write("cp compute_MI_dd_orderness.py %s/BB/PHI/ORDERNESS/.\n"%fil_name)
                
                f_nc.write("cp compute_MI_dd_orderness.py %s/BB/PSI/ORDERNESS/.\n"%fil_name)
                f_nc.write("cp compute_MI_dd_orderness.py %s/BB/PHI/ORDERNESS/.\n"%fil_name)
                
                f_nc.write("cp compute_MI_dd_orderness.py %s/SC/CHI1/ORDERNESS/.\n"%fil_name)
                
                
f_nc.close()                
                
```
