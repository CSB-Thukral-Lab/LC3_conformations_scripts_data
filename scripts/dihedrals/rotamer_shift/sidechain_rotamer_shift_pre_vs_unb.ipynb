```python

def rot_frac(system_rot_all):
    
    system_all_sims_rot_states=[]
    res_ls_system=[]

    for resid_ind in range(system_rot_all.shape[1]):
        res_ls_system.append(system_rot_all.columns[resid_ind])

        #SC HAS G+, TRANS AND G- rotamer states

        #close
        rotameric_states_ls=system_rot_all[system_rot_all.columns[resid_ind]].to_list()
        f_gauche_min_close=rotameric_states_ls.count(0)/len(rotameric_states_ls)
        f_trans_close=rotameric_states_ls.count(1)/len(rotameric_states_ls)
        f_gauche_max_close=rotameric_states_ls.count(2)/len(rotameric_states_ls)
        system_all_sims_rotameric_states_l=[f_gauche_min_close,f_trans_close,
                                  f_gauche_max_close]
        system_all_sims_rot_states.append(system_all_sims_rotameric_states_l)

    system_all_sims_SC_rotameric_states=pd.DataFrame()
    system_all_sims_SC_rotameric_states['RESIDUES']=res_ls_system
    system_all_sims_SC_rotameric_states['f(G-)']=[i[0] for i in system_all_sims_rot_states]
    system_all_sims_SC_rotameric_states['f(Trans)']=[i[1] for i in system_all_sims_rot_states]
    system_all_sims_SC_rotameric_states['f(G+)']=[i[2] for i in system_all_sims_rot_states]
    
    return system_all_sims_SC_rotameric_states

def num_ext(DF, angle_type):
    FORM_dih_sel_col=[]
    for i in DF['RESIDUES']:
        j=i.split('.')[0].replace('chi%d'%angle_type,'')
        t_ls=re.findall(r'\d+', j)
        t=int(''.join(t_ls))
        FORM_dih_sel_col.append(t)
    return(FORM_dih_sel_col)

def rotamer_shift(f1_rotamers, f2_rotamers):
    df1=abs(f1_rotamers[0]-f2_rotamers[0])
    df2=abs(f1_rotamers[1]-f2_rotamers[1])
    df3=abs(f1_rotamers[2]-f2_rotamers[2])
    
    change=(df1+df2+df3)/3
    
    return(change)

def rot_shift_compute(pre_rot_state_frac, unb_rot_state_frac, angle_type_): # pre, unb
    F_shift=[]
    for i in range(1,120):
        if i in num_ext(unb_rot_state_frac, angle_type_):
            PRE=pre_rot_state_frac[pre_rot_state_frac['Rn']==i]
            PRE_ROT=[PRE['f(G-)'].iloc[0],PRE['f(Trans)'].iloc[0],PRE['f(G+)'].iloc[0]]

            UNB=unb_rot_state_frac[unb_rot_state_frac['Rn']==i]
            UNB_ROT=[UNB['f(G-)'].iloc[0],UNB['f(Trans)'].iloc[0],UNB['f(G+)'].iloc[0]]

            F_shift.append(rotamer_shift(PRE_ROT,UNB_ROT))

        else:
            F_shift.append(0)
    
    F_shift_df=pd.DataFrame()
    F_shift_df['Residues']=[i for i in range(1,120)]
    F_shift_df['Shift']=F_shift
    return F_shift_df
```

```python
CHI1_pre_rot_fracr=rot_frac(CHI1_pre_rot_all)
CHI2_pre_rot_fracr=rot_frac(CHI2_pre_rot_all)
CHI3_pre_rot_fracr=rot_frac(CHI3_pre_rot_all)
CHI4_pre_rot_fracr=rot_frac(CHI4_pre_rot_all)


CHI1_unb_rot_fracr=rot_frac(CHI1_unb_rot_all)
CHI2_unb_rot_fracr=rot_frac(CHI2_unb_rot_all)
CHI3_unb_rot_fracr=rot_frac(CHI3_unb_rot_all)
CHI4_unb_rot_fracr=rot_frac(CHI4_unb_rot_all)
```

```python
CHI1_pre_rot_fracr['Rn']=num_ext(CHI1_pre_rot_fracr, 1)
CHI2_pre_rot_fracr['Rn']=num_ext(CHI2_pre_rot_fracr, 2)
CHI3_pre_rot_fracr['Rn']=num_ext(CHI3_pre_rot_fracr, 3)
CHI4_pre_rot_fracr['Rn']=num_ext(CHI4_pre_rot_fracr, 4)

CHI1_unb_rot_fracr['Rn']=num_ext(CHI1_unb_rot_fracr, 1)
CHI2_unb_rot_fracr['Rn']=num_ext(CHI2_unb_rot_fracr, 2)
CHI3_unb_rot_fracr['Rn']=num_ext(CHI3_unb_rot_fracr, 3)
CHI4_unb_rot_fracr['Rn']=num_ext(CHI4_unb_rot_fracr, 4)

```

```python
chi1_shift_pre_vs_unb=rot_shift_compute(CHI1_pre_rot_fracr, CHI1_unb_rot_fracr, 1)
chi2_shift_pre_vs_unb=rot_shift_compute(CHI2_pre_rot_fracr, CHI2_unb_rot_fracr, 2)
chi3_shift_pre_vs_unb=rot_shift_compute(CHI3_pre_rot_fracr, CHI3_unb_rot_fracr, 3)
chi4_shift_pre_vs_unb=rot_shift_compute(CHI4_pre_rot_fracr, CHI4_unb_rot_fracr, 4)
```

```python
chi_shift_select=[]
for i in range(1,120):
    chi1_fil=chi1_shift_pre_vs_unb[chi1_shift_pre_vs_unb['Residues']==i]['Shift'].to_list()[0]
    chi2_fil=chi2_shift_pre_vs_unb[chi2_shift_pre_vs_unb['Residues']==i]['Shift'].to_list()[0]
    chi3_fil=chi3_shift_pre_vs_unb[chi3_shift_pre_vs_unb['Residues']==i]['Shift'].to_list()[0]
    chi4_fil=chi4_shift_pre_vs_unb[chi4_shift_pre_vs_unb['Residues']==i]['Shift'].to_list()[0]
    chi_shift_select.append([chi1_fil, chi2_fil, chi3_fil, chi4_fil])
    
    
```

```python
chi_shift_select_df=pd.DataFrame(chi_shift_select)
chi_shift_select_df.columns=['CHI1','CHI2','CHI3','CHI4' ]
chi_shift_select_df['RESIDUES']=[i for i in range(1,120)]

chi_shift_select_df
```
