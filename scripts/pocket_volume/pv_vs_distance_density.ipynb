```python

#density calculations
def manhattan_dist_1d(p1,p2):
    manhattan_dist=abs(p1-p2)
    return(manhattan_dist)

def chebyshev_dist(p1,p2):
    return(max(np.abs(p1[0]-p2[0]),np.abs(p1[1]-p2[1])))

def neigh_ret(k_val, df_n):
    k3_all=k_val


    xy=[]
    for i,j in zip(df_n['NET VOL'],
                df_n['NET MEAN DIST']):
        xy.append([i/max(df_n['NET VOL']),j/max(df_n['NET MEAN DIST'])])

    neigh=[]
    for i in xy:
        counter=0
        ref_point=i
        radius=[]
        for j in xy:
            if j!=ref_point:
                dist_measure=chebyshev_dist(ref_point,j)
                if dist_measure<k3_all:
                    counter=counter+1

        neigh.append(counter) 
    
    return neigh
```

```python
NEIGH_PRE=neigh_ret(0.05, PV_DIST_VOL_PRE)
NEIGH_UNB=neigh_ret(0.05, PV_DIST_VOL_UNB)
NEIGH_BND=neigh_ret(0.05, PV_DIST_VOL_BND)
```
