```python
def manhattan_dist_1d(p1,p2):
    manhattan_dist=abs(p1-p2)
    return(manhattan_dist)

def chebyshev_dist(p1,p2):
    return(max(np.abs(p1[0]-p2[0]),np.abs(p1[1]-p2[1])))

def neigh_ret(k_val, df_n,col_name):
    k3_all=k_val
    xy=[]
    for i in tqdm.tqdm(df_n[col_name]):
        xy.append(i/max(df_n[col_name]))
        neigh=[]
        for i in xy:
            counter=0
            ref_point=i
            radius=[]
            for j in xy:
                if j!=ref_point:
                    dist_measure=manhattan_dist_1d(ref_point,j)
                    if dist_measure<k3_all:
                        counter=counter+1

            neigh.append(counter) 
    
    return neigh
```

```python
NEIGH_LS=[]
for r_i, col_ in zip(HP1, HP1_PRE_dist_dat.T):
    #print(r_i, col_)
    tl_df=HP1_PRE_dist_dat.T[[col_]]
    NEIGH_test=neigh_ret(0.03, tl_df, col_)
    max_neigh_ind=NEIGH_test.index(max(NEIGH_test))
    NEIGH_LS.append(NEIGH_test)
    
    print(r_i,tl_df.index[max_neigh_ind], tl_df[col_].to_list()[max_neigh_ind])
```
